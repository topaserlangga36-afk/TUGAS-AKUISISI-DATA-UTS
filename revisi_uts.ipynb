```python
import zipfile
import os

zip_path = '/content/TUGAS-AKUISISI-DATA-UTS-REVISI.zip'
extract_path = '/content/extracted_data'

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_path)

# List the extracted files
for root, dirs, files in os.walk(extract_path):
    for file in files:
        print(os.path.join(root, file))
```

```python
import json

readme_path = '/content/extracted_data/TUGAS-AKUISISI-DATA-UTS-main/README.md'
notebook_path = '/content/extracted_data/TUGAS-AKUISISI-DATA-UTS-main/notebooks/akuisisi_data_uts.ipynb'

print("=== README.md Content ===")
try:
    with open(readme_path, 'r', encoding='utf-8') as f:
        print(f.read())
except Exception as e:
    print(f"Error reading README: {e}")

print("\n=== Notebook Structure ===")
try:
    with open(notebook_path, 'r', encoding='utf-8') as f:
        nb_content = json.load(f)
    print(f"Number of cells in notebook: {len(nb_content.get('cells', []))}")
    for idx, cell in enumerate(nb_content.get('cells', [])[:15]): # Print first 15 cell previews
        cell_type = cell.get('cell_type')
        source = "".join(cell.get('source', []))[:120].replace('\n', ' ')
        print(f"Cell {idx} ({cell_type}): {source}...")
except Exception as e:
    print(f"Error reading Notebook: {e}")
```
